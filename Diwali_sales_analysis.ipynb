EDA(Exploratory Data)

```python
import seaborn as sns
```

```python
df.columns
```

```python
sns.countplot(x='Gender',data=df)
```

```python
ax=sns.countplot(x='Gender',data=df)
for bar in ax.containers:
     ax.bar_label(bar)
print(ax.containers)
```

```python
df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
## it arrange the values by groupby 
```

```python
sales_gen=df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x='Gender', y='Amount', data=sales_gen)
```

Most of the buyers are female and purchase power greater than male

```python
sns.countplot(data=df, x="Age Group",hue="Gender")
```

```python
sales_age = df.groupby(["Age Group"],as_index=False)["Amount"].sum().sort_values(by='Amount',ascending=False)
print(sales_age)
```

```python
sns.barplot(x="Age Group",y="Amount",data=sales_age, hue="Age Group")
```

Occupation analysis

```python
sns.set(rc={'figure.figsize':(18,5)})
ax=sns.countplot(data=df, x="Occupation", hue='Occupation')
# print(ax.containers)
for bar in ax.containers:
#   print(ax.bar_label(bar))
  ax.bar_label(bar)
```

```python
ocuupation_ = df.groupby(["Occupation"],as_index=False)["Amount"].sum().sort_values(by='Amount',ascending=False)
# print(ocuupation_)
sns.set(rc={"figure.figsize":(16,6)})
sns.barplot(x="Occupation",y="Amount",data=ocuupation_, hue="Occupation")
```

"from the above graph we can see that most of the buyers are from IT sector"

```python
sns.set(rc={'figure.figsize':(20,5)})
ax=sns.countplot(data=df, x="Product_Category", hue='Product_Category')
# print(ax.containers)
for bar in ax.containers:
#   print(ax.bar_label(bar))
  ax.bar_label(bar)
```
